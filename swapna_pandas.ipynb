### Q1. Array Operations and Vectorization

#### Given Code

The following code creates a NumPy array and a Pandas Series containing the same values. The operation `* 2` is then applied to both structures.

```python
import numpy as np
import pandas as pd

arr = np.array([10, 20, 30])
s = pd.Series([10, 20, 30])

print(arr * 2)
print(s * 2)
```

#### Predicted Output

**NumPy Array:**

```text
[20 40 60]
```

**Pandas Series:**

```text
0    20
1    40
2    60
dtype: int64
```

#### Operation Insight

The multiplication by `2` is a **vectorized operation**. Instead of applying the multiplication separately to each value using a loop, NumPy and Pandas apply the operation directly to the complete data structure.

#### Key Observation

Both structures produce the same numerical values after multiplication. The main difference in the output is that the **Pandas Series keeps an index** (`0, 1, 2`) along with the values, while the NumPy array displays only the values.

#### Similarity

Both NumPy arrays and Pandas Series support **element-wise arithmetic operations**, so each value is multiplied by `2`.

#### Difference

A **NumPy array** is mainly used for efficient numerical and array-based computation, whereas a **Pandas Series** stores values together with an index, making it useful for working with labeled data.

#### Conclusion

The result shows how **vectorization allows the same operation to be performed efficiently across all elements without writing an explicit loop**.

## Q1. NumPy Broadcasting

### Given Code

The two NumPy arrays have different shapes: `a` has shape `(3, 1)` and `b` has shape `(1, 3)`. NumPy uses **broadcasting** to make their shapes compatible for element-wise addition.

### Array Shapes Before Addition

```text
a → (3, 1)
b → (1, 3)
```

### Expected Output

```text
[[11 12 13]
 [21 22 23]
 [31 32 33]]
(3, 3)
```

### Operation Insight

Broadcasting allows NumPy to virtually expand the smaller dimensions so that the two arrays can be added without manually creating a larger array.

The `(3, 1)` array is matched with the `(1, 3)` array, producing a final shape of `(3, 3)`.

### Why the Result Is 3 × 3

The values in `a` are arranged vertically:

```text
10
20
30
```

The values in `b` are arranged horizontally:

```text
1  2  3
```

Broadcasting combines each value from `a` with every value from `b`.

### Edge Case: If `b = np.array([1, 2])`

The addition still works because NumPy can broadcast the shape `(2,)` across the second dimension of the `(3, 1)` array.

The resulting shape becomes `(3, 2)`:

```text
[[11 12]
 [21 22]
 [31 32]]

### Conclusion

Broadcasting allows arrays with compatible shapes to participate in element-wise operations without explicitly reshaping or copying the data.

````python
import numpy as np

# 'a' is arranged vertically with 3 rows and 1 column.
# 'b' is arranged horizontally with 1 row and 3 columns.
# These shapes allow NumPy to use broadcasting during addition.
a = np.array([[10],
              [20],
              [30]])

b = np.array([[1, 2, 3]])

# Broadcasting expands the two arrays virtually to the same
# shape and performs element-wise addition.
result = a + b

print("Array a:")
print(a)

print("\nShape of a:")
print(a.shape)

print("\nArray b:")
print(b)

print("\nShape of b:")
print(b.shape)

print("\nResult after broadcasting and addition:")
print(result)

print("\nShape of the result:")
print(result.shape)

# This case is intentionally tested separately because
# the new shape cannot be broadcast with a (3, 1) array.
b_invalid = np.array([1, 2])

print("\nTrying an incompatible shape:")

try:
    print(a + b_invalid)
except ValueError as error:
    print("ValueError:", error)

````

## Q2. CSV Parsing, Missing Values and Date Handling

### Given Dataset

The CSV contains a quoted comma inside the `Name` column, missing marks, and dates written in different formats.

### Parsing Observation

The value `"Asha, R"` contains a comma, but because the name is enclosed in double quotes, Pandas treats it as **one field** instead of splitting it into two columns.

### Potential Issues

The `Date` column contains different date formats, so it should be converted into a consistent datetime format.

The `Marks` column contains both a blank value and `"NA"`. These values should be investigated before treating them as missing marks.

### Robust Approach

The data should be loaded normally, the `Date` column should be converted using `pd.to_datetime()`, and the `Marks` column should be converted to numeric values so that invalid or missing entries can be identified.

### Why `"NA"` Needs Investigation

`"NA"` may represent a missing mark, but it could also be a value used by the original data source for another reason. It should therefore be investigated before automatically treating it as a missing value.

### Conclusion

Correct CSV parsing is important because an incorrectly parsed column can change the structure of the DataFrame and lead to incorrect analysis.

````python
import pandas as pd

# Keep values such as "NA" unchanged initially so that
# their meaning can be investigated before treating them as missing.
df = pd.read_csv("students.csv", keep_default_na=False)

print("Original DataFrame:")
display(df)

print("\nOriginal Data Types:")
display(df.dtypes)

# Convert dates into a consistent datetime representation.
df["Date"] = pd.to_datetime(df["Date"], dayfirst=True, errors="coerce")

# Convert Marks to numeric values.
# Invalid entries such as "NA" become missing values (NaN).
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")

print("\nDataFrame after cleaning Date and Marks:")
display(df)

print("\nUpdated Data Types:")
display(df.dtypes)
````

## Q3. Salary Outlier and Robust Statistics

### Given Data

The salary column contains three values in a relatively small range and one extremely large value. This makes it important to compare the **mean** and **median** before deciding which statistic represents a typical salary.

### Predicted Output

**Mean:**

```text
2522500.0
```

**Median:**

```text
32500.0
```

### Effect of the Outlier

The salary `10000000` is much larger than the other salaries. Since the mean uses every value in the calculation, this extreme value pulls the mean strongly upward.

The median is much less affected because it depends on the middle position after sorting the values.

### Better Representation of a Typical Salary

For this dataset, the **median** better represents a typical employee salary because it is not strongly affected by the extreme salary value.

### Important Observation

The extreme value should not be removed automatically. It may be a genuine salary belonging to a senior employee, or it could be a data-entry error.

Therefore, the record should be **investigated before removing it**.

### Conclusion

The mean is sensitive to extreme values, while the median is more robust when the dataset contains an outlier.

````python
import pandas as pd

# Create the salary data so we can compare the effect
# of the unusually large salary on mean and median.
df = pd.DataFrame({
    "Employee": ["A", "B", "C", "D"],
    "Salary": [25000, 30000, 35000, 10000000]
})

print("Salary Data:")
display(df)

# Mean considers every salary, so the extreme value
# has a strong effect on the final average.
mean_salary = df["Salary"].mean()

# Median uses the middle position after sorting,
# making it less sensitive to an extreme value.
median_salary = df["Salary"].median()

print("\nMean Salary:", mean_salary)
print("Median Salary:", median_salary)

# Identify the employee record with the highest salary
# so the unusual value can be investigated.
highest_salary = df.iloc[df["Salary"].to_numpy().argmax()]

print("\nHighest Salary Record:")
print(highest_salary)
````

# Section D – Scenario-Based Problems

## Q1. Student Assessment Data Cleaning

### Problem Understanding

The dataset contains different types of data-quality issues such as inconsistent department names, duplicate records, missing marks, text values in the `Marks` column, and marks outside the valid range.

The cleaning process should handle these issues step by step while preserving useful information for investigation.

### Cleaning Sequence

1. Load and inspect the dataset.
2. Standardize department names.
3. Identify and remove exact duplicate records.
4. Preserve and investigate the original `Marks` entries.
5. Convert the `Marks` column into a numeric format.
6. Distinguish between absent students and marks that have not been entered.
7. Identify marks outside the valid `0–100` range.
8. Display the cleaned DataFrame.
9. Report the cleaning results.

### Handling Missing Marks

A blank mark, `Absent`, and `NA` should not automatically be treated as the same situation.

A blank value can indicate that the mark has not been entered, while `Absent` indicates that the student did not attend the assessment. The `NA` value should also be investigated before deciding how it should be treated.

### Risk of Deleting Missing Values

Deleting all records with missing marks without investigating them can remove valid student information and hide the actual reason why the data is missing.

### Handling Invalid Marks

Marks outside the valid range of `0–100` should be identified and investigated before deciding whether they should be corrected or removed.

For example, a mark of `105` may indicate a data-entry error rather than a valid assessment result.

````python
import pandas as pd

# Load the dataset so that the original records can be
# inspected before making any cleaning changes.
df = pd.read_csv("student_assessment_dirty.csv")

print("===== INITIAL DATASET =====")
display(df)

print("\n===== FIRST 5 RECORDS =====")
display(df.head())

print("\n===== DATASET INFORMATION =====")
df.info()

print("\n===== MISSING VALUES =====")
display(df.isnull().sum().to_frame("Missing Count"))


# Standardize department names so that different spellings
# such as IT, it and I.T. are represented consistently.
df["Department"] = (
    df["Department"]
    .str.strip()
    .str.upper()
    .replace({
        "I.T.": "IT"
    })
)

print("\n===== STANDARDIZED DEPARTMENTS =====")
display(df[["Student_ID", "Department"]].drop_duplicates())


# Identify exact duplicate records before removing them.
duplicate_count = df.duplicated().sum()

print("\n===== EXACT DUPLICATE RECORDS =====")
display(df[df.duplicated(keep=False)])

print("Number of exact duplicate records:", duplicate_count)

# Remove only exact duplicate records.
df = df.drop_duplicates()


# Preserve the original Marks values before converting them.
# This allows us to distinguish Absent, NA and blank entries.
df["Marks_Status"] = df["Marks"].astype("string").str.strip()

df["Marks_Status"] = df["Marks_Status"].replace({
    "Absent": "Absent",
    "NA": "Not Available"
})

# A blank value means the mark has not been entered.
df.loc[
    df["Marks_Status"].isna() | (df["Marks_Status"] == ""),
    "Marks_Status"
] = "Not Entered"


print("\n===== MARKS STATUS BEFORE NUMERIC CONVERSION =====")
display(
    df[
        ["Student_ID", "Student_Name", "Marks", "Marks_Status"]
    ]
)


# Remove the word "marks" from values such as "88 marks".
# The remaining values can then be converted into numbers.
df["Marks"] = (
    df["Marks"]
    .astype("string")
    .str.strip()
    .str.replace(r"\s*marks\s*$", "", regex=True, case=False)
)

# Convert valid numeric values into numbers.
# Values such as Absent and NA become NaN during conversion.
df["Marks"] = pd.to_numeric(df["Marks"], errors="coerce")


print("\n===== MISSING MARKS AFTER CONVERSION =====")
display(
    df[
        df["Marks"].isna()
    ][
        ["Student_ID", "Student_Name", "Marks_Status", "Marks"]
    ]
)


# Identify marks outside the valid assessment range of 0–100.
invalid_marks = (
    df["Marks"].notna()
    & ~df["Marks"].between(0, 100)
)

print("\n===== INVALID MARKS OUTSIDE 0–100 =====")
display(
    df.loc[
        invalid_marks,
        ["Student_ID", "Student_Name", "Marks"]
    ]
)


# Count the results required for the final report.
missing_count = df["Marks"].isna().sum()
invalid_count = invalid_marks.sum()


print("\n===== CLEANED DATASET =====")
display(df)


# Create a simple summary table for the final cleaning report.
report = pd.DataFrame({
    "Measure": [
        "Records removed as exact duplicates",
        "Missing marks remaining",
        "Invalid marks identified"
    ],
    "Count": [
        duplicate_count,
        missing_count,
        invalid_count
    ]
})

print("\n===== CLEANING REPORT =====")
display(report)
````
